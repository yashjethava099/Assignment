### Write two prompts for ChatGPT: one asking for a list of trending songs in India (unclear), and one asking for a list of trending Bollywood songs released after 2022 with artist names (clear). Compare the responses and note the difference in detail and relevance.

Prompt 1 — Unclear

“Give me a list of trending songs in India.”

Likely response:
A broad list covering different languages, industries, and genres—such as Bollywood, Punjabi, Telugu, Tamil, and independent music. The definition of “trending” and the time period may also be unclear.

Prompt 2 — Clear

“Give me a list of trending Bollywood songs released after 2022. Include the song title, artist/singer name, movie name, and release year. Focus on songs that have been popular in India recently.”

Likely response:
A more focused list of Bollywood songs released from 2023 onward, with useful details such as song title, singer/artist, movie, and release year. The response can also be limited to songs relevant to recent Indian trends.

Comparison
Aspect	Prompt 1	Prompt 2
Scope	Very broad	Specifically Bollywood
Time period	Not specified	After 2022
Artist information	Not requested	Explicitly requested
Other details	Not specified	Movie + release year
Relevance	May include unrelated languages/genres	Much more targeted
Response detail	Basic/broad	Detailed and structured

Conclusion: The second prompt produces a more relevant, precise, and detailed response because it clearly defines the music category, release period, and information required. The first prompt is ambiguous, so ChatGPT has to make assumptions about what “trending songs” means.
