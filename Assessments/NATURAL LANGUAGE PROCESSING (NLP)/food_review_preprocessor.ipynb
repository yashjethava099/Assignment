# Food Review Text Preprocessor

A reusable text preprocessing pipeline for cleaning raw food delivery customer reviews, ready for downstream NLP tasks.

**Pipeline steps:**
1. Lowercase the text
2. Remove punctuation
3. Strip extra whitespace
4. Tokenise
5. Remove English stopwords (NLTK)
6. Apply stemming (PorterStemmer)
7. Return a single space-joined string


```python
# Install dependencies (uncomment if running for the first time)
# !pip install nltk
```

```python
import re
import string
import nltk

# Download required NLTK resources (only needs to run once)
nltk.download('stopwords', quiet=True)
nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)

from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
from nltk.tokenize import word_tokenize

STOPWORDS = set(stopwords.words('english'))
stemmer = PorterStemmer()
```

## Preprocessing Function

```python
def preprocess_review(text: str) -> str:
    """
    Clean and preprocess a raw food delivery customer review for NLP tasks.

    Steps:
        1. Lowercase the text
        2. Remove punctuation
        3. Strip extra whitespace
        4. Tokenise
        5. Remove English stopwords
        6. Apply Porter stemming
        7. Join tokens back into a single string

    Parameters
    ----------
    text : str
        Raw review text.

    Returns
    -------
    str
        Cleaned, stemmed, space-joined string of tokens.
    """
    if not isinstance(text, str) or not text.strip():
        return ""

    # 1. Lowercase
    text = text.lower()

    # 2. Remove punctuation
    text = text.translate(str.maketrans('', '', string.punctuation))

    # Remove digits (optional but common for review cleaning) - keep numbers if you need them
    # text = re.sub(r'\d+', '', text)

    # 3. Strip extra whitespace
    text = re.sub(r'\s+', ' ', text).strip()

    # 4. Tokenise
    tokens = word_tokenize(text)

    # 5. Remove stopwords
    tokens = [tok for tok in tokens if tok not in STOPWORDS]

    # 6. Stem
    tokens = [stemmer.stem(tok) for tok in tokens]

    # 7. Join back into a single string
    return " ".join(tokens)
```

## Test on Sample Food Delivery Reviews

```python
sample_reviews = [
    "The food arrived COLD and 45 minutes late!!! Very disappointed, won't be ordering again.",
    "Amazing biryani, super fast delivery, and the packaging was neat & mess-free. Highly recommend!!",
    "Order was missing 2 items, and when I called customer support they weren't helpful at all...",
]

for i, review in enumerate(sample_reviews, start=1):
    cleaned = preprocess_review(review)
    print(f"Review {i} (raw):     {review}")
    print(f"Review {i} (cleaned): {cleaned}")
    print("-" * 80)
```

## Optional: Lemmatization Variant

If you prefer lemmatization (context-aware, dictionary-based) over stemming (rule-based truncation), use `WordNetLemmatizer` instead.

```python
nltk.download('wordnet', quiet=True)
nltk.download('omw-1.4', quiet=True)

from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

def preprocess_review_lemmatized(text: str) -> str:
    """Same pipeline as preprocess_review, but uses lemmatization instead of stemming."""
    if not isinstance(text, str) or not text.strip():
        return ""

    text = text.lower()
    text = text.translate(str.maketrans('', '', string.punctuation))
    text = re.sub(r'\s+', ' ', text).strip()

    tokens = word_tokenize(text)
    tokens = [tok for tok in tokens if tok not in STOPWORDS]
    tokens = [lemmatizer.lemmatize(tok) for tok in tokens]

    return " ".join(tokens)


# Test the lemmatized version on the same samples
for i, review in enumerate(sample_reviews, start=1):
    cleaned = preprocess_review_lemmatized(review)
    print(f"Review {i} (lemmatized): {cleaned}")
```
