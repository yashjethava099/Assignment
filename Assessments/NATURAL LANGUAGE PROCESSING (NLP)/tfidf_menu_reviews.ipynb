# TF-IDF Vectoriser for Menu Reviews

Apply TF-IDF vectorisation to a corpus of food delivery menu item reviews and identify the most informative terms per document.

**Steps:**
1. Build a corpus of at least 8 short reviews (positive, negative, neutral)
2. Vectorise with `TfidfVectorizer`
3. Display the full TF-IDF matrix as a labelled `pandas` DataFrame
4. Print the top 3 highest-scoring words per review


```python
# Install dependencies (uncomment if running for the first time)
# !pip install scikit-learn pandas
```

```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer

pd.set_option('display.max_columns', 50)
pd.set_option('display.width', 150)
```

## Corpus of Menu Item Reviews

```python
corpus = [
    "The butter chicken was rich, creamy, and absolutely delicious.",         # positive
    "Delivery was late and the pizza arrived cold and soggy.",                # negative
    "The pasta was okay, nothing special but not bad either.",                # neutral
    "Best biryani I have ever had, packed with flavour and perfectly spiced.",# positive
    "Portion size was tiny for the price, very disappointing meal.",          # negative
    "The sushi rolls were fresh and beautifully presented.",                  # positive
    "Service was average, the burger was just a regular fast food burger.",   # neutral
    "Cold noodles, bland sauce, and a rude delivery driver ruined the order.",# negative
    "The dessert was decent, sweet but a little too sugary for my taste.",    # neutral
    "Fantastic curry with authentic spices, will definitely order again!",    # positive
]

for i, review in enumerate(corpus):
    print(f"[{i}] {review}")
```

## TF-IDF Vectorisation

```python
vectorizer = TfidfVectorizer(stop_words='english', lowercase=True)
tfidf_matrix = vectorizer.fit_transform(corpus)

feature_names = vectorizer.get_feature_names_out()

tfidf_df = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=feature_names,
    index=[f"Review {i}" for i in range(len(corpus))]
)

tfidf_df
```

## Top 3 Terms per Review

```python
def top_n_terms(row, n=3):
    """Return the top-n (term, score) pairs for a single TF-IDF row, sorted descending by score."""
    nonzero = row[row > 0]
    top = nonzero.sort_values(ascending=False).head(n)
    return list(top.items())


for idx, review in enumerate(corpus):
    row = tfidf_df.iloc[idx]
    top_terms = top_n_terms(row, n=3)
    print(f"Review {idx}: \"{review}\"")
    for term, score in top_terms:
        print(f"    {term:<15} {score:.4f}")
    print("-" * 70)
```

## Notes

- `TfidfVectorizer` was configured with `stop_words='english'` to remove common English stopwords before scoring, so the most informative (content-bearing) words rise to the top.
- Words that are rare across the corpus but frequent within a specific review get the highest TF-IDF scores — these are typically the most distinctive/descriptive words for that review (e.g. `"creamy"`, `"soggy"`, `"biryani"`).
- Words appearing in many reviews (e.g. `"food"`, `"order"`) get down-weighted by the IDF component, even if frequent.
