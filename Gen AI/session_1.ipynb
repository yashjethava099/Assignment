## 2. Discriminative Models vs Generative Models

| Aspect             | Discriminative Models                                                 | Generative Models                                                    |                      |
| ------------------ | --------------------------------------------------------------------- | -------------------------------------------------------------------- | -------------------- |
| Definition         | Learn the boundary between classes and predict labels from input data | Learn the underlying data distribution and generate new data samples |                      |
| Main Use           | Classification and prediction tasks                                   | Data generation, content creation, and simulation                    |                      |
| Example Task       | Email spam detection                                                  | Generating realistic emails or text                                  |                      |
| Example Model Name | Logistic Regression, SVM, Random Forest                               | GPT, GAN, Variational Autoencoder (VAE)                              |                      |
| Output             | Predicts a label or category                                          | Creates new text, images, audio, or code                             |                      |
| Learning Goal      | Learn P(Y                                                             | X)                                                                   | Learn P(X) or P(X,Y) |


## 3. Product Description Generation Using ChatGPT
Prompt Used

Write a creative product description for a new Flipkart gadget called "SmartBand X1". It is a lightweight fitness band with AI-powered health monitoring, 14-day battery life, sleep tracking, heart-rate monitoring, water resistance, and a 1.8-inch AMOLED display. Make the description attractive for online shoppers and keep it under 120 words.

Output Received

SmartBand X1 – Your Smart Fitness Companion

Stay ahead of your fitness goals with the SmartBand X1, designed to keep you connected and healthy every day. Featuring a vibrant 1.8-inch AMOLED display, AI-powered health insights, advanced heart-rate monitoring, and detailed sleep tracking, it helps you understand your body better. With a long-lasting 14-day battery and water-resistant design, SmartBand X1 is ready for workouts, adventures, and everyday life. Lightweight, stylish, and packed with smart features, it's the perfect companion for anyone looking to improve their health while staying connected on the go.

## 4. New Generative AI Feature for an Existing App
App: Spotify
Proposed Feature: AI Music Story Creator

I would add a feature called "My Music Journey" to Spotify.

How it works:

The generative AI model analyzes the user's listening history, favorite artists, moods, and playlists.
It generates a personalized weekly story describing the user's music journey.
The AI can create summaries such as:
"This week you explored energetic workout tracks and discovered 5 new artists."
"Your music mood shifted from relaxing acoustic songs to upbeat pop."

Generative AI Technology Used:

A text-generation model (such as GPT) would analyze listening patterns and generate natural-language summaries.
This would make music analytics more engaging and personalized compared to traditional charts and statistics.

Benefit:
Users receive meaningful insights about their listening habits in a fun, human-like narrative rather than just numbers and graphs.
